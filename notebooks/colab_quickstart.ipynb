## 4. Look inside: one hand-written backward pass

```python
import numpy as np
from nnbench.numpy_nn import Conv2D
conv = Conv2D(1, 4, 3, padding=1)
x = np.random.randn(2, 1, 8, 8).astype(np.float32)
y = conv.forward(x)
dx = conv.backward(np.ones_like(y))
print(y.shape, dx.shape, conv.grads['W'].shape)
```
